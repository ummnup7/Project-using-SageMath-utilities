```python
# Heat Conduction from a Sphere - Part (a)
# Solve differential equation using desolve() - Symbolic Display
from sage.all import *

# Define symbolic variables
r = var('r', domain='positive')
a, T0, T_inf = var('a T_0 T_infty', domain='positive')
T = function('T')(r)
print("=" * 80)
print("HEAT CONDUCTION FROM A SPHERE")
print("=" * 80)

print("\nLaplace equation in spherical coordinates (radial symmetry):")
print("∇²T = 0")

print("\nIn radial form:")
print("(1/r²) d/dr(r² dT/dr) = 0")
print("\nMultiplying by r²:")
diff_eq = diff(r**2 * diff(T, r), r) ==0
print(f"d/dr(r² dT/dr) = 0")

print(f"\nDifferential equation to solve:")
show(diff_eq)
```

```python
general_solution = desolve(diff_eq, T)

print(f"\n {T} = ")
show(general_solution)
```

```python
# Step 3: Apply boundary conditions
print("\n" + "Apply Boundary Conditions")
print("-" * 80)

print("\nBoundary Condition 1 (BC1):")
print("At r = a (sphere surface): T = T₀")
print(f"T(a) = T₀")

print("\nBoundary Condition 2 (BC2):")
print("As r → ∞ (far field): T → T∞")
print(f"lim(r→∞) T(r) = T∞")
```

```python
# Extract constants from general solution
C1, C2 = symbols('_K1 _K2', real=True)
print("\nGeneral form:")
print(f"T(r) = {C1} + {C2}/r")

print("\n--- From BC2 (r → ∞) ---")
print(f"lim(r→∞) [{C1} + {C2}/r] = T∞")
print(f"{C2}/r → 0 as r → ∞")
print(f"Therefore: {C1} = T∞")
C1_sol = T_inf
```

```python
print("\n--- From BC1 (r = a) ---")
print(f"At r = a: {C1_sol} + {C2}/a = T₀")
print(f"T₀ = T∞ + {C2}/a")
print(f"Solving for {C2}:")
print(f"{C2} = (T₀ - T∞) × a")
C2_sol = (T0 - T_inf) * a
print(f"\nConstants:")
print(f"{C1} = {C1_sol}")
print(f"{C2} = {C2_sol}")
```

```python

T_final = C1_sol + C2_sol/r
T_final_simplified = simplify(T_final)

print(f"\nSubstituting constants back:")
print(f"T(r) = {C1_sol} + {C2_sol}/r")

print(f"\nSimplified:")
print(f"T(r) = {T_final_simplified}")

print(f"\n" + "=" * 80)
print("FINAL RESULT FOR PART (a):")
print("=" * 80)
print(f"\nT(r) = T∞ + (T₀ - T∞) × (a/r)")
print(f"\nor equivalently:")
print(f"\nT(r) = {T_final_simplified}")

```

```python
var('r a T0 T_inf beta theta')

# (a) Monopole solution
T_a = T_inf + (T0 - T_inf)*a/r
disturbance_a = T_a - T_inf
print("Monopole soln, T=")
show(T_a)

# (b) Dipole solution
T_b = T_inf + beta*r*cos(theta)*(1 + (a^3)/(2*r^3))
background_b = T_inf + beta*r*cos(theta)
disturbance_b = T_b - background_b
print("Dipole soln, T=")
show(T_b)

# Simplify
disturbance_a_simple = disturbance_a.simplify_full()
disturbance_b_simple = disturbance_b.simplify_full()

print("Disturbance for part (a):")
show(disturbance_a_simple)

print("Disturbance for part (b):")
show(disturbance_b_simple)

# Get the leading order term as r -> infinity using series expansion
print("Series expansion for large r (part a):")
series_a = disturbance_a_simple.series(r==oo, 2)
show(series_a)

print("Series expansion for large r (part b):")
series_b = disturbance_b_simple.series(r==oo, 2)
show(series_b)
```

```python
# Define symbols
var('r a T0 T_inf beta theta')

# Part (a): monopole disturbance term
T_a = T_inf + (T0 - T_inf)*a/r
disturbance_a = T_a - T_inf  # Disturbance = (T0 - Tinf)*a/r
print("Part (a) disturbance:")
show(disturbance_a)

# Analyze decay rate by multiplying disturbance by r
decay_check_a = simplify(r * disturbance_a)
print("Multiplying disturbance by r gives (should be constant for 1/r decay):")
show(decay_check_a)

# Part (b): dipole disturbance term
T_b = T_inf + beta*r*cos(theta)*(1 + a^3/(2*r^3))
background_b = T_inf + beta*r*cos(theta)
disturbance_b = T_b - background_b  # Disturbance = (beta a^3/2r^2) cos(theta)
print("Part (b) disturbance:")
show(disturbance_b)

# Analyze decay rate by multiplying disturbance by r^2
decay_check_b = simplify(r^2 * disturbance_b)
print("Multiplying disturbance by r^2 gives (should be constant for 1/r^2 decay):")
show(decay_check_b)

# Explanation as comments in code
print('''\
- For part (a), r * disturbance = constant means disturbance ~ 1/r (monopole decay).
- For part (b), r^2 * disturbance = constant means disturbance ~ 1/r^2 (dipole decay).
''')
```

```python
var('r a theta T_inf beta k lambda_')

# Define symbols for constants of integration inside (A, B) and outside (C, D)
A, B, C, D = var('A B C D')

# Axisymmetric solution form using Legendre polynomial P1:
# Outside solution:
# T_out(r, theta) = T_inf + beta * r * cos(theta) + C * cos(theta)/r**2
T_out = T_inf + beta * r * cos(theta) + C * cos(theta)/r^2
print("T out=")
show(T_out)

# Inside solution (finite at r=0):
# T_in(r, theta) = A * r * cos(theta)
T_in = A * r * cos(theta)
print("T in=")
show(T_in)

```

```python
# Define variables and parameters
var('r a theta T_inf beta k lambda_ A C')

# Define temperature fields
T_out = T_inf + beta * r * cos(theta) + C * cos(theta)/r^2
T_in = A * r * cos(theta)

# Boundary condition 1: Temperature continuity at r = a
eq1 = T_in.subs(r=a) == T_out.subs(r=a)

# Boundary condition 2: Heat flux continuity at r = a
dT_in_dr = diff(T_in, r)
dT_out_dr = diff(T_out, r)

bc2_left = lambda_ * k * dT_in_dr.subs(r=a)
bc2_right = k * dT_out_dr.subs(r=a)

eq2 = bc2_left == bc2_right

# Solve for A, C
sol = solve([eq1, eq2], A, C, solution_dict=True)[0]

A_sol = sol[A]
C_sol = sol[C]

print("Solution for constants:")
print(f"A = {A_sol}")
print(f"C = {C_sol}")

# Substitute constants back into temperature fields
T_in_sol = T_in.subs(A=A_sol).simplify_full()
T_out_sol = T_out.subs(C=C_sol).simplify_full()

print("\nTemperature inside the sphere:")
show(T_in_sol)

print("Temperature outside the sphere:")
show(T_out_sol)

# Heat flux at r = a outside the sphere
q_r = -k * diff(T_out_sol, r)
q_r_surface = q_r.subs(r=a).simplify_full()

print("\nHeat flux at the sphere surface (outside):")
show(q_r_surface)

# Surface area of sphere
A_sphere = 4 * pi * a^2

# Total heat flow Q = integral of heat flux over surface
# The heat flux q_r has angular dependence cos(theta), integrate over sphere:
# Integral of q_r*cos(theta)*sin(theta) dtheta dphi = 0 because cos(theta) is antisymmetric about equator
# But we perform integral for verification using symbolic integration

# Surface integral over theta and phi:
phi = var('phi')
integrand = q_r_surface * a^2 * sin(theta)  # Surface element dS = a^2 sin(theta) dtheta dphi

total_Q = integral(integral(integrand, theta, 0, pi), phi, 0, 2*pi).simplify_full()

print("\nTotal heat flux from the sphere:")
show(total_Q)
```
